```python
# Download first N seconds of a video (<=720p) into video/data/raw
import yt_dlp
from yt_dlp.utils import download_range_func

video_url = 'https://www.ted.com/talks/matt_cutts_try_something_new_for_30_days'
clip_seconds = 60

assert video_url, "set video_url first"

opts = {
    "format": "bv*[height<=720]+ba/b[height<=720]/b",  # not ext-specific: TED serves mp4 audio
    "download_ranges": download_range_func(None, [(0, clip_seconds)]),
    "outtmpl": "video/data/raw/%(id)s.%(ext)s",
    "compat_opts": {"no-certifi"},  # use system CA store (TMA corporate CA)
}
with yt_dlp.YoutubeDL(opts) as ydl:
    ydl.download([video_url])
```
