```python
model = WhisperModel(
    "small",
    device="cpu",
    compute_type="int8",
)
```

```python
segments, info = model.transcribe(
    "data/raw/32-34_audio.1c239b86.mp3"
)

segments = list(segments)

print(info)

audio_segments = [
    {
        "segment_id": segment.id,
        "start": segment.start,
        "end": segment.end,
        "text": segment.text.strip(),
        "avg_logprob": segment.avg_logprob,
        "no_speech_prob": segment.no_speech_prob,
    }
    for segment in segments
]

audio_segments
```

```python
import json

output_path = "data/processed/32-34_audio_segments.json"

with open(output_path, "w", encoding="utf-8") as f:
    json.dump(audio_segments, f, ensure_ascii=False, indent=2)

print(f"Saved {len(audio_segments)} segments to {output_path}")
```
